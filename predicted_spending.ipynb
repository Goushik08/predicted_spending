```python
import pandas as pd
```

```python
dataset =pd.read_csv("customer_data.csv")
```

```python
dataset
```

```python
dataset = dataset.drop(columns=['name','country'])
```

```python
dataset
```

```python
dataset =pd.get_dummies(dataset,drop_first=True)
```

```python
dataset
```

```python
dataset.columns
```

```python
x = dataset[['age', 'income', 'purchase_frequency','gender_Male',
       'education_High School', 'education_Master', 'education_PhD']]
```

```python
x
```

```python
y=dataset[[ 'spending']]
```

```python
y
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=11
)
```

```python
from sklearn.linear_model import LinearRegression 
regressor=LinearRegression() 
regressor.fit(x_train,y_train)
```

```python
y_pred=regressor.predict(x_test) 
y_pred
```

```python
x_test
```

```python
y_test
```

```python
from sklearn.metrics import r2_score 
percentage=r2_score(y_test,y_pred)*100
print(f"result: {percentage} %")
```

```python
import pickle

filename = "spending.sav"

pickle.dump(regressor, open(filename, "wb"))


loaded_model = pickle.load(open(filename, "rb"))
```

```python
age = int(input("Enter age: "))

gender = input("Enter gender (Male/Female): ").lower()

education = input("Enter education (High School/Bachelor/Master/PhD): ").lower()

income = float(input("Enter income: "))

purchase_frequency = float(input("Enter purchase frequency: "))

if gender == "male":
    gender_male = 1
else:
    gender_male = 0


if education == "high school":
    education_high_school = 1
    education_master = 0
    education_phd = 0

elif education == "master":
    education_high_school = 0
    education_master = 1
    education_phd = 0

elif education == "phd":
    education_high_school = 0
    education_master = 0
    education_phd = 1

elif education == "bachelor":
    education_high_school = 0
    education_master = 0
    education_phd = 0

else:
    print("Invalid education")
    exit()
import pandas as pd
prediction_input = pd.DataFrame([[
    age,
    income,
    purchase_frequency,
    gender_male,
    education_high_school,
    education_master,
    education_phd
]], columns=[
    "age",
    "income",
    "purchase_frequency",
    "gender_Male",
    "education_High School",
    "education_Master",
    "education_PhD"
])


result = loaded_model.predict(prediction_input)

predicted_spending = result[0][0]

print(f"For {age} year old {gender.title()} with {education.title()} education spending is {predicted_spending:.2f}")
```
